# 03 — Modelos quânticos (janelas IMDC)

QRC, QSVM, ensemble por regime (QSVM→RF), VQR Re-uploading, VQR AngleEmbedding e QLSTM. Os resultados foram computados pelo `run_estudo_imdc.py` (execução longa); aqui carregamos os JSON salvos e exibimos as figuras.

```python
import matplotlib; matplotlib.use('Agg')
import os, sys, json, warnings; warnings.filterwarnings('ignore')
import numpy as np, pandas as pd
from IPython.display import Image, display
AQUI = os.path.abspath('..')
sys.path.insert(0, AQUI)
FIG = os.path.join(AQUI, 'figuras'); RES = os.path.join(AQUI, 'resultados')
import imdc_splits as S
quant=['QRC','QSVM','EnsRegime','VQR_ReUp','VQR_Angle','QLSTM']
tab=[]
for nome in quant:
    p=os.path.join(RES,f'{nome}_resultados.json')
    if not os.path.exists(p): print('faltando:',nome); continue
    d=json.load(open(p,encoding='utf-8'))
    for c in ['T1','T2','T3']:
        if c in d: tab.append([nome,c,round(d[c].get('R2',float('nan')),3),round(d[c].get('WIS',float('nan')),0)])
pd.DataFrame(tab, columns=['Modelo','Teste','R2','WIS'])
```

```python
for nome in quant:
    fp=os.path.join(FIG, f'{nome}_pred_vs_obs.png')
    if os.path.exists(fp): display(Image(fp))
```
